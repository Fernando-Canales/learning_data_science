## Z-test

### The expression for a Z-score is given by:

## $ Z =  \frac{\bar{x} - \mu}{\frac{\sigma}{\sqrt{N}}} $

### where $\mu$ is the mean of the population, $\bar{x}$ is the mean of the sample, $\sigma$ is the standard deviation of the population and $\rm n$ is the sample size

### A confidence interval for the population mean in terms of Z-score is defined by:
## $ \mu = \bar{x} \pm Z \cdot \left( \frac{\sigma}{\sqrt{n}} \right) $

#### Ex. 1.  Past studies have shown that the mean recovery time for a cold is 18 days, with a standard deviation of 1.5 days, and follows a normal distribution. Now let’s say an experimental new drug was given to a test group of 40 people, and it took an average of 16 days for them to recover from the cold. Did the drug have an impact?

````python
# Answer

mu = 18 # population mean
std = 1.5 # population std

n = 40 # sample size
x = 16 # sample mean

alpha = 0.05 # statistical significance

#1. State hypothesis
"""
H0: mu <= 18
H1: mu < 18
"""

#2. Compute the Z-scores
z_x = (x - mu)/(std/np.sqrt(n))

#3. Compute the p-value for the Z-score
p_value = st.norm.cdf(z_x)


print("Z-score is:", z_x)
print("p-value is:", p_value)

#4. Check hypothesis
if p_value < alpha:
    print("We reject H0, the drug did have an impact!")
else:
    print("We can't reject H0, the drug did not have any impact!")
````

#### As we saw, in this case, since the population was normally distributed, we performed what is known as a ```Z-test```. Obtained the p-values directly from Z-scores. However, for different cases, we might use t-tests. I know that ```t-tests```are extremely useful and usually a nice alternative to ```Z-tests```.

## t-test

### The t-score or 't-statistic' is defined as follow

## $ t = \frac{\bar{x} - \mu}{\frac{s}{\sqrt{n}}}$

### where 's' is the standard deviation of the sample (t-tests are used when we don't know the standard deviation of the population '$\sigma$')
### The confidence interval for the population mean when performing a t-test is given by:
## $ \mu = \bar{x} \pm t \cdot \frac{s}{\sqrt{n}}$

#### Ex. 1.  Past studies have shown that the mean recovery time for a cold is 18 days, with a standard deviation of 1.5 days, and follows a normal distribution. Now let’s say an experimental new drug was given to a test group of 40 people, and it took an average of 16 days for them to recover from the cold. Did the drug have an impact?

````python
# Simulate sample data (40 people taking the drug)
np.random.seed(42)
sample_data = np.random.normal(loc=16, scale=1.5, size=40)

# Perform one-sample t-test with one-sided alternative
t_stat, p_value = st.ttest_1samp(a=sample_data, popmean=18, alternative='two-sided')

print(f"T-Statistic: {t_stat:.2f}")
print(f"P-Value: {p_value:.10f}")

alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis. The drug significantly reduced recovery time.")
else:
    print("Fail to reject the null hypothesis. No significant impact.")
````

### Python has very nice libraries, likes scipy and scipy has a thing called 'stats'. And there we can use `ttest_1samp()` is for comparing a sample mean to a known population mean and we can use `ttest_ind()` for comparing the means of two independent groups (i.e. the sample means of two samples).

**1. What is ```scipy.stats.ttest_1samp()```?**
- **Purpose**: Compares the **mean of a sample with the mean of the population** (e.g. test if a drug changes recovery time with respect to a knownn population).
- **Use Case**: When you have **a population and a sample** and want to compare their means


**2. What is `scipy.stats.ttest_ind()`?**
- **Purpose**: Compares the **means of two independent groups** (e.g., test if a drug
changes recovery time compared to a control group).
- **Use Case**: When you have **two groups** (e.g., 40 people taking the drug vs. 40 people
not taking the drug) and want to compare their **sample means**.


### **`scipy.stats.ttest_ind()` is for comparing **two independent samples**, not a sample vs. a population.**

#### **Key Use Case**:
- **Two groups**: e.g., a drug group and a control group (no drug).
- **Goal**: Test whether the **means of the two groups differ** (e.g., does the drug change
recovery time compared to the control?).

#### **Example**:
Simulate you have two groups of people. Each group has 40 participants. To one group (control group) you don't give any drug or you give a placebo to konw what is the mean recovery time of a cold. To the other group yo do give a specific drug to see what is the mean reco

````python
from scipy import stats
import numpy as np

# Simulate two groups: drug group and control group
np.random.seed(42)
drug_group = np.random.normal(loc=16, scale=1.5, size=40)  # Drug group
control_group = np.random.normal(loc=18, scale=1.5, size=40)  # Control group

# Perform two-sample t-test
t_stat, p_value = stats.ttest_ind(drug_group, control_group)
print("The t-score is:", t_stat)
print("The p-value is:", p_value)

# Check the hypothesis
alpha = 0.05
if p_value <= alpha:
    print("We reject H0: The drug did have an impact!")
else:
    print("We can't reject H0: The drug did not have any impact!")
````
