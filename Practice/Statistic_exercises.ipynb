#### Ex 1: 
You bought a spool of 1.75 mm filament for your 3D printer. You want to
measure how close the filament diameter really is to 1.75 mm. You use a caliper
tool and sample the diameter five times on the spool:
1.78, 1.75, 1.72, 1.74, 1.77
Calculate the mean and standard deviation for this set of values.

```python
# Answer
measurements = [1.78, 1.75, 1.72, 1.74, 1.77]

def obtain_mean(values):
    mean = sum(values)/len(values)
    return mean
```

```python
def obtain_variance(values):
    mean = sum(values)/len(values)

    numerator = 0
    for i in range(len(values)):
        numerator = numerator + (values[i] - mean)**2
    variance = numerator/len(values)
    return variance
```

```python
def obtain_standard_deviation(variance):
    std = np.sqrt(variance)
    return std
```

```python
mean = obtain_mean(measurements)
print("The mean is:", mean)
```

```python
variance = obtain_variance(measurements)
print("The variance is:", variance)
```

```python
std = obtain_standard_deviation(variance)
print("The standard deviation is:", std)
```

#### Ex. 2:
A manufacturer says the Z-Phone smart phone has a mean consumer life of 42
months with a standard deviation of 8 months. Assuming a normal distribution,
what is the probability a given random Z-Phone will last between 20 and 30
months?

```python
#Answer
mean = 42
std = 8

upper_limit = norm.cdf(30, mean, std) 
lower_limit = norm.cdf(20, mean, std)

answer = upper_limit - lower_limit
```

```python
print("The probability is:", answer, "and in percentage (%)", answer*100,"%")
```

#### Ex. 3:
I am skeptical that my 3D printer filament is not 1.75 mm in average diameter
as advertised. I sampled 34 measurements with my tool. The sample mean is
1.715588 and the sample standard deviation is 0.029252.

What is the 99% confidence interval for the mean of my entire spool of filament?

```python
# Answer

"""
To answer this question we have to:
1. Obtain the Z-scores
2. Obtain the confidence interval with the formula of the estimate, E = +-z*sample_std/sqrt(sample_size)
"""

#1.
def obtain_z_scores(p):
    standard_normal_distribution = norm(loc=0.0, scale=1.0)
    left_area = (1.0 - p)/2.0
    right_area = 1.0 - ((1.0 - p)/2.0)
    return norm.ppf(left_area), norm.ppf(right_area) # inverse CDF to obtain Z-scores, this is a tuple

#2.
def confidence_interval(z_left, z_right, sample_mean, sample_std, sample_size):
    lower_ci = z_left*(sample_std/np.sqrt(sample_size))
    upper_ci = z_right*(sample_std/np.sqrt(sample_size))
    return sample_mean + lower_ci, sample_mean + upper_ci # this is alsoa  tuple

n = 34 # sample_size
x = 1.71588 # sample mean
s = 0.029252 # sample std

z_score_left, z_score_right = obtain_z_scores(p=0.99)
ci = confidence_interval(z_score_left, z_score_right, x, s, n)

print("Z-scores are:", z_score_left, ",", z_score_right)
print("The confidence interval is:", ci)
```

#### Ex. 5:
Your marketing department has started a new advertising campaign and wants to
know if it affected sales, which in the past averaged 10,345 a day with a standard deviation of 552. The new advertising campaign ran for 45 days and averaged $11,641 in sales. Did the campaign affect sales? Why or why not? (Use a two-tailed test for more
reliable significance.)

```python
# Answer

"""
To answer this question we have to:
1. Define the H0 and the H1
2. Define a statistical significane threshold (alpha)
3. Define the standard error (s/sqrt(n)) and other parameters
4. Obtain the two x-values for each tail to define the rejection zone
5. Obtain the two p-values (or the single p-value)
5. Compare the p-value with the value of alpha
"""

#1.
# H0 -> mu_before = mu_after 
# H1 -> mu_before != mu_after

#2.
alpha = 0.05 # statistical significance

#3.
mean_population = 10345
std_population = 552
mean_sample = 11641
sample_size = 45
standard_error = std_population/np.sqrt(sample_size)
# Now, following the philosphy of the book, we define the distance between the population mean and the sample mean to also have a mirror number before the population mean
distance = abs(mean_population - mean_sample) # 1296

number_before_the_population_mean = mean_population - distance  # 10345 - 1296 = 9049
sample_mean = mean_population + distance # 10345 + 1296 = 11641 (see variable mean_sample from above)

#4. We obtain here the rejection zone (from which x-values our statistical significance exists)
def obtain_x_values(alpha):
    x1 = norm.ppf(alpha/2, mean_population, std_population)
    x2 = norm.ppf(1 - alpha/2, mean_population, std_population)
    return x1, x2 # this is a tuple

#5. Obtaine the 2 p-values, one p-value for the sample mean and the other p-value for the number we obtained  in step 3. (the distance)
def obtain_p_values(left_boundary, right_boundary):
    # Obtain the p-value of the mean being as extreme than the left tail
    p1 = norm.cdf(left_boundary, mean_population, standard_error) # p-value of the sample mean
    # Obtain the p-value of the mean being as extreme than the right tail
    p2 = 1.0 - norm.cdf(right_boundary, mean_population, standard_error) # p-value of the second or right tail
    return p1, p2 # this is a tuple

#5.
x_values = obtain_x_values(0.05)
p_values = obtain_p_values(number_before_the_population_mean, sample_mean)

print("The x-values are:", x_values)
print("The p-values are:", p_values, "and their sum is:", sum(p_values))
print(" ")
print(" ")
print("--------------Hypothesis testing now---------------")
print(" ")
print(" ")
if sum(p_values) <= alpha:
    print("We reject H0, then the campaign had an impact!")
else:
    print("We cannot reject H0, then the campaing didn't have any impact!")
print(" ")
print(" ")
print("--------------Hypothesis testing now---------------")
```

#### Let's solve the same problem in an easier way

```python
population_mean = 10345
population_standard_deviation = 552
sample_mean = 11641
sample_size = 45
alpha = 0.05

# Now, let's compute the standard error
standard_error = population_standard_deviation / np.sqrt(sample_size)

# Now, let's obtain the distance between the population mean and the sample mean
distance = abs(population_mean - sample_mean) # 1296
print("The distance is:", distance)

# Now let's find a mirror number behind the population mean (the mirror of the sample mean)
left_number = population_mean - distance
print("The mirror number is:", left_number)

# Now let's obtain the Z-score of the sample mean and of the mirror number
z_sample_mean = (sample_mean - population_mean) / standard_error
z_left_number = (left_number - population_mean) / standard_error

print("The Z-score of the left number is:", z_left_number)
print("The Z-score of the sample mean is:", z_sample_mean)


# Now let's obtain the p-value for this z-score (it corresponds to the right tail)
p_value_sample_mean = 1.0 - norm.cdf(z_sample_mean)

print("p-value of the sample mean:", p_value_sample_mean)

# Now let's obtain the p-valye of the mirror number
p_value_left_number = norm.cdf(z_left_number)
print("p-value of the mirror number:", p_value_left_number)

p_value = p_value_sample_mean + p_value_left_number
print("total p-value:", p_value)


# Let's check the hypothesis
if p_value < alpha:
    print("Let's reject H0. There was an actual influence and change.")
else:
    print("We can't reject H0. There's no change")
```
