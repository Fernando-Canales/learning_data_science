Logistic regression is a useful machine learning algorithm for classifying things. The output of a Logistic Regression is always an integer between 0 and 1. That means, the output of a logistic regression can be seen as a probability.

In Linear Regression we learned to fit a straight line to a data set, in a Logistic Regression what we fit is a S-shape curve or a ```Sigmoid``` that is expressed as:

$$
y = \frac{1}{1 - e^{-(\beta_0 + x \beta_1)}}
$$

or as (equivalently):

$$
y = \frac{e^{(\beta_0 + x \beta_1)}}{1 - e^{(\beta_0 + x \beta_1)}}
$$

Logistic regression is extremely useful for defining if a given data point could be inside or outside a given category. It is also useful to define thresholds or walls between groups of data points.

Let's dive in the MATH!

## 1 Logistic Function

We could write the sigmoid or S-shape curve as follow

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

If we plug our linear equation into the sigmoid, our predicted probabilty  becomes

$$
\hat{y} = \frac{1}{1 + e^{-(m x + b)}}
$$

Using the matrix/vector versions of the equations, as we did for the Linear Regression, we standarize our derviation in terms of how it is written in Machine Learning. We vectorize our parameters $\rm m$ and $\rm b$ into the vecotr $\boldsymbol{\theta}$ and our input data set into the vector $\mathbf{X}$ with an extra 1 to manage the multupliciation

$$
\boldsymbol{\theta} = \begin{bmatrix}
\theta_0 \\
\theta_1
\end{bmatrix} =
\begin{bmatrix}
b \\
m
\end{bmatrix}, \mathbf{x} =\begin{bmatrix} 
1 \\
x
\end{bmatrix}
$$

The linear combination $m \cdot x + b$ can then be written as 

$$
\boldsymbol{\theta}^{T} \cdot \mathbf{x} = \begin{bmatrix}
b & m
\end{bmatrix} \cdot
\begin{bmatrix}
1 \\
x
\end{bmatrix}
$$

Our prediction for a single data point is then

$$
\hat{y}^{(i)} = \sigma(\boldsymbol{\theta}^{T} \cdot \mathbf{x}^{(i)}) \frac{1}{1 + e^{-(\boldsymbol{\theta}^{T} \cdot \mathbf{x}^{(i)})}}
$$

## 2 Maximun Likelihood Estimation

In Linear Regression, we minimized the Mean Squared Error (MSE). If we use MSE with the sigmoid function, the resulting error surface is non-convex—it looks like a terrain with multiple valleys (local minima), which destroys Gradient Descent.

Instead, we use probability. For a single observation, we want to maximize the probability that our prediction $\hat{y}^{(i)}$ matches the true label $y^{(i)} \in \{0,1\}$:
- If $y^{(i)} = 1$, we want $\hat{y}^{(i)}$ to be close to $1$.
- If $y^{(i)} = 0$, we want $(1 - \hat{y}^{(i)})$ to be close to $1$ (that is $\hat{y}^{(i)} = 0$)

We can condense these two conditions into a single Bernoulli probability expression:$$P(y^{(i)} \vert{} \mathbf{x}^{(i)}, \boldsymbol{\theta}) = (\hat{y}^{(i)})^{y^{(i)}} (1 - \hat{y}^{(i)})^{1 - y^{(i)}}$$Assuming all $N$ data points are independent, the total probability of observing our entire dataset given our parameters $\boldsymbol{\theta}$ is the product of all individual probabilities. This is the Likelihood Function, $L(\boldsymbol{\theta})$:$$L(\boldsymbol{\theta}) = \prod\limits_{i=1}^{N} (\hat{y}^{(i)})^{y^{(i)}} (1 - \hat{y}^{(i)})^{1 - y^{(i)}}$$

Maximum Likelihood Estimation dictates that we must find the parameters $\boldsymbol{\theta}$ that maximize $L(\boldsymbol{\theta})$.

## 3 Log-Likelihood to Cost Function (NLL)

Multiplying thousands of small probabilities results in numerical underflow (computers struggle to store values that close to zero). Furthermore, taking the derivative of a massive multiplication chain is a nightmare.To solve this, we take the natural logarithm of the Likelihood function. Because logarithms are monotonically increasing, maximizing the Log-Likelihood will yield the exact same $\boldsymbol{\theta}$ as maximizing the raw Likelihood. The logarithm brilliantly turns the giant product into a giant sum:

$$\ell(\boldsymbol{\theta}) = \log L(\boldsymbol{\theta}) = \sum_{i=1}^{N} \left[ y^{(i)} \log(\hat{y}^{(i)}) + (1 - y^{(i)}) \log(1 - \hat{y}^{(i)}) \right]$$

Optimization algorithms like Gradient Descent are conventionally built to minimize error, not maximize it. Therefore, we multiply the Log-Likelihood by $-\frac{1}{N}$ to create the Negative Log-Likelihood (NLL), also known as the Binary Cross-Entropy Cost Function, $J(\boldsymbol{\theta})$:

$$J(\boldsymbol{\theta}) = -\frac{1}{N} \sum_{i=1}^{N} \left[ y^{(i)} \log(\hat{y}^{(i)}) + (1 - y^{(i)}) \log(1 - \hat{y}^{(i)}) \right]
$$

## 4 Performing the derivates to obtain an expression for the gradient descent

To perform Gradient Descent, we need the gradient of our cost function: $\nabla_{\boldsymbol{\theta}} J$. We must compute the partial derivative of $J$ with respect to  $b$ and $m$.

We define our linear equation and our prediction for a single data point $i$:

$$
z = m \cdot x + b
$$ 

$$
\hat{y} = \sigma(z) = \frac{1}{1 + e^{-z}}
$$

The cost function for a single data point (omitting the sum and $1/N$ for a moment to focus on the calculus) is:
$$
J = - \left[ y \log(\hat{y}) + (1 - y) \log(1 - \hat{y}) \right]
$$

To find how the cost changes when we tweak $m$ or $b$, we use the chain rule. The error depends on the prediction $\hat{y}$, which depends on the linear output $z$, which depends on our parameters $m$ and $b$:


$$
\frac{\partial J}{\partial m} = \frac{\partial J}{\partial \hat{y}} \cdot \frac{\partial \hat{y}}{\partial z} \cdot \frac{\partial z}{\partial m}
$$

$$
\frac{\partial J}{\partial b} = \frac{\partial J}{\partial \hat{y}} \cdot \frac{\partial \hat{y}}{\partial z} \cdot \frac{\partial z}{\partial b}
$$

Step 1: The derivative of the Cost ($J$) with respect to the prediction ($\hat{y}$)

$$
\frac{\partial J}{\partial \hat{y}} = - \left[ y \left( \frac{1}{\hat{y}} \right) + (1 - y) \left( \frac{1}{1 - \hat{y}} \right) (-1) \right]
$$

$$
\frac{\partial J}{\partial \hat{y}} = - \frac{y}{\hat{y}} + \frac{1 - y}{1 - \hat{y}} = \frac{-y(1 - \hat{y}) + \hat{y}(1 - y)}{\hat{y}(1 - \hat{y})} = \frac{-y + y\hat{y} + \hat{y} - y\hat{y}}{\hat{y}(1 - \hat{y})} = \frac{\hat{y} - y}{\hat{y}(1 - \hat{y})}
$$

Step 2: The derivative of the prediction ($\hat{y}$) with respect to the linear output ($z$)This is the derivative of the sigmoid function, which evaluates to the function multiplied by one minus the function:

$$
\frac{\partial \hat{y}}{\partial z} = \hat{y}(1 - \hat{y})
$$

Step 3: The derivative of the linear output ($z$) with respect to $m$ and $b$ Since $z = m \cdot x + b$:

$$
\frac{\partial z}{\partial m} = x
$$

$$
\frac{\partial z}{\partial b} = 1
$$

Step 4: Putting the Chain Rule back togetherFor the slope $m$:

$$
\frac{\partial J}{\partial m} = \left( \frac{\hat{y} - y}{\hat{y}(1 - \hat{y})} \right) \cdot \left( \hat{y}(1 - \hat{y}) \right) \cdot (x)
$$

The denominators cancel out beautifully:

$$
\frac{\partial J}{\partial m} = (\hat{y} - y) \cdot x
$$

For the intercept $b$:

$$
\frac{\partial J}{\partial b} = \left( \frac{\hat{y} - y}{\hat{y}(1 - \hat{y})} \right) \cdot \left( \hat{y}(1 - \hat{y}) \right) \cdot (1)
$$
Again, the complex fractions cancel:

$$
\frac{\partial J}{\partial b} = (\hat{y} - y)
$$

When we expand this across the entire dataset of $N$ points to get our final average gradients, we arrive at the exact formulas used to update the weights in Python:

$$
\frac{\partial J}{\partial m} = \frac{1}{N} \sum_{i=1}^{N} (\hat{y}^{(i)} - y^{(i)}) \cdot x^{(i)}
$$

$$
\frac{\partial J}{\partial b} = \frac{1}{N} \sum_{i=1}^{N} (\hat{y}^{(i)} - y^{(i)})
$$

Arriving to the expression for the gradient for all the dataset:

$$
\nabla_{\theta} \mathbf{J} =  \begin{bmatrix}
\frac{\partial J}{\partial b} \\
\frac{\partial J}{\partial m}
\end{bmatrix} = -\frac{1}{N} \sum_{i=1}^{N} (\hat{y}^{(i)} - y^{(i)}) \cdot \begin{bmatrix}
1  \\
x^{(i)}
\end{bmatrix}
$$

Getting into more detail:

$$
\sum_{i=1}^{N} (\hat{y}^{(i)} - y^{(i)}) \begin{bmatrix} 1 \\ x^{(i)} \end{bmatrix} = (\hat{y}^{(1)} - y^{(1)}) \begin{bmatrix} 1 \\ x^{(1)} \end{bmatrix} + (\hat{y}^{(2)} - y^{(2)}) \begin{bmatrix} 1 \\ x^{(2)} \end{bmatrix} + \dots + (\hat{y}^{(N)} - y^{(N)}) \begin{bmatrix} 1 \\ x^{(N)} \end{bmatrix}
$$

Now, let's look at the matrices we defined earlier:The Design Matrix $\mathbf{X}$ (shape $N \times 2$), where every row is a data point vector $\mathbf{x}^{(i)^T} = \begin{bmatrix} 1 & x^{(i)} \end{bmatrix}$:

$$
\mathbf{X} = \begin{bmatrix} 1 & x^{(1)} \\  1 & x^{(2)} \\  \vdots & \vdots \\  1 & x^{(N)} \end{bmatrix}
$$

Therefore, its Transpose $\mathbf{X}^T$ (shape $2 \times N$) flips the columns into rows:

$$
\mathbf{X}^T = \begin{bmatrix} 1 & 1 & \dots & 1 \\  x^{(1)} & x^{(2)} & \dots & x^{(N)} \end{bmatrix}
$$

The Error Vector (shape $N \times 1$):
$$
(\hat{\mathbf{y}} - \mathbf{y}) = \begin{bmatrix} \hat{y}^{(1)} - y^{(1)} \\  \hat{y}^{(2)} - y^{(2)} \\  \vdots \\  \hat{y}^{(N)} - y^{(N)} \end{bmatrix}
$$

Watch what happens when you perform the matrix multiplication $\mathbf{X}^T (\hat{\mathbf{y}} - \mathbf{y})$:

$$
\mathbf{X}^T (\hat{\mathbf{y}} - \mathbf{y}) = \begin{bmatrix} 1 & 1 & \dots & 1 \\  x^{(1)} & x^{(2)} & \dots & x^{(N)} \end{bmatrix} \begin{bmatrix} \hat{y}^{(1)} - y^{(1)} \\  \hat{y}^{(2)} - y^{(2)} \\  \vdots \\  \hat{y}^{(N)} - y^{(N)} \end{bmatrix} = \begin{bmatrix} 1 \cdot (\hat{y}^{(1)} - y^{(1)}) + 1 \cdot (\hat{y}^{(2)} - y^{(2)}) + \dots + 1 \cdot (\hat{y}^{(N)} - y^{(N)}) \\  x^{(1)} \cdot (\hat{y}^{(1)} - y^{(1)}) + x^{(2)} \cdot (\hat{y}^{(2)} - y^{(2)}) + \dots + x^{(N)} \cdot (\hat{y}^{(N)} - y^{(N)}) \end{bmatrix}$$

For the resulting $2 \times 1$ vector: The top row is the sum of all errors: $\sum (\hat{y}^{(i)} - y^{(i)})$, which is $\frac{\partial J}{\partial b}$ (multiplied by $N$).The bottom row is the sum of errors weighted by $x$: $\sum x^{(i)}(\hat{y}^{(i)} - y^{(i)})$, which is $\frac{\partial J}{\partial m}$ (multiplied by $N$).Thus, matrix multiplication handles the entire summation loop natively. Scaling it down by $\frac{1}{N}$ gives you the final compact expression:$$\nabla_{\boldsymbol{\theta}} J = \frac{1}{N} \mathbf{X}^T (\hat{\mathbf{y}} - \mathbf{y})$$

This has the same shape as the expression for the closed gradient descent for the linear regression.

````python
# let's define the sigmoid

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# let's plot it just for fun
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 100)

sigma = sigmoid(x)

plt.plot(x, sigma)
````
